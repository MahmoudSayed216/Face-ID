```python
from face_embedder.FaceEmbedder import FaceEmbedder
import os

# input_base = "input"
images = os.listdir("input")


embedder = FaceEmbedder(detection_model='YOLO26NANO_BASE', embedding_model='r18')

images = [os.path.join("input", image) for image in images]

images
```

```python
embeddings = embedder.embed_faces_from_desk(images)
```

```python
len(list(embeddings.keys()))
```

```python
import numpy as np

vectors, face_imgs, metadata = [], [], []
for img_path, faces in embeddings.items():
    for i, (vec, face_arr) in enumerate(faces):
        vectors.append(vec)
        face_imgs.append(face_arr)
        metadata.append((img_path, i))

X = np.array(vectors, dtype=np.float32)
X = X / np.linalg.norm(X, axis=1, keepdims=True)

print(f"Faces: {len(X)} | Dim: {X.shape[1]} | Images: {len(embeddings)}")
```

```python
from sklearn.manifold import TSNE

n = len(X)
perplexity = min(30, max(2, n // 3))

tsne = TSNE(n_components=2, perplexity=perplexity,
            metric='cosine', init='pca', random_state=42, n_iter_without_progress=10000)
X_2d = tsne.fit_transform(X)
print(f"Done. perplexity={perplexity}")
```

```python
import plotly.graph_objects as go
from ipywidgets import Output, HBox
from IPython.display import display
import matplotlib.pyplot as plt
import os

# color per image
img_paths = [p for p, _ in metadata]
unique_imgs = list(dict.fromkeys(img_paths))
color_map = {p: i for i, p in enumerate(unique_imgs)}
colors = [color_map[p] for p in img_paths]

hover_text = [f"{os.path.basename(p)} · face {i}" for p, i in metadata]

fig = go.FigureWidget(data=[go.Scatter(
    x=X_2d[:, 0], y=X_2d[:, 1],
    mode='markers',
    marker=dict(size=10, color=colors, colorscale='Rainbow',
                line=dict(width=1, color='black')),
    text=hover_text,
    hoverinfo='text',
)])
fig.update_layout(width=700, height=600,
                  title=f't-SNE of {len(X)} faces — hover to see crop')

image_out = Output()

def on_hover(trace, points, state):
    if not points.point_inds:
        return
    idx = points.point_inds[0]
    with image_out:
        image_out.clear_output(wait=True)
        face = face_imgs[idx]
        plt.figure(figsize=(2.5, 2.5))
        plt.imshow(face)                    # add [..., ::-1] if BGR
        plt.axis('off')
        plt.title(hover_text[idx], fontsize=9)
        plt.show()

fig.data[0].on_hover(on_hover)
display(HBox([fig, image_out]))
```
